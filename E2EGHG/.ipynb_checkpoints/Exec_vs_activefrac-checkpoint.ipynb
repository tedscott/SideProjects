## Compute my own summary stats from the trace_energy_totals csv

```python
df_raw
```

```python
df_summary = df_raw.drop(['trace_id','zone','kepler_duration_sum_ms','start_energy_joules',
                         'end_energy_joules','energy_start_ts','energy_end_ts','exper_name',
                         'interval_ms','sleep_ms','idle_scrapping_ms','source_csv'], axis=1)
df_summary
```

```python
df_mean = df_summary.groupby(['cpu_ms','active_interval_ms'], as_index=False).mean()
df_mean=df_mean.rename(columns={'energy_joules':'mean_energy_joules'})
df_std = df_summary.groupby(['cpu_ms','active_interval_ms'], as_index=False).std()
df_std=df_std.rename(columns={'energy_joules':'std_energy_joules'})
#print(df_mean, df_std)
```

```python
# df_temp = df_summary.groupby(['cpu_ms','active_interval_ms'], as_index=False).agg(["mean","std"])
# print(df_temp)
df_summarised = pd.DataFrame([df_mean.cpu_ms,df_mean.active_interval_ms,df_mean.mean_energy_joules,df_std.std_energy_joules]).transpose()
df_summarised
```

```python
df_summarised['polling_freq'] = 1000 / df_summarised.active_interval_ms
df_summarised
```

```python
df_summarised.query("cpu_ms == 250 & active_interval_ms == 25")
```

```python
df_spans_agg = df_spans.drop(['sample_number','timestamp','exper_name','interval_ms',
                         'sleep_ms','idle_scrapping_ms','exper_name','source_csv','cpu_ratio','energy_ratio'], axis=1)
df_spans_agg
```

```python
df_spans_agg_2 = df_spans_agg.groupby(['trace_id','cpu_ms','active_interval_ms'], as_index=False).sum()
df_spans_agg_2 = df_spans_agg_2.drop(['trace_id'], axis=1)
df_spans_agg_2['cpu_ratio'] = df_spans_agg_2.kepler_duration_sample/df_spans_agg_2.cpu_ms
df_spans_agg_2['energy_ratio'] = df_spans_agg_2.kepler_energy_sample/df_spans_agg_2.energy_sample
df_spans_agg_2

df_spans_mean = df_spans_agg_2.groupby(['cpu_ms','active_interval_ms'], as_index=False).mean()
df_spans_mean=df_spans_mean.rename(columns={'kepler_duration_sample':'kepler_duration_sample_mean', 'kepler_energy_sample':'kepler_energy_sample_mean',
                                            'energy_sample':'energy_sample_mean','cpu_ratio':'cpu_ratio_mean','energy_ratio':'energy_ratio_mean'})
df_spans_std = df_spans_agg_2.groupby(['cpu_ms','active_interval_ms'], as_index=False).std()
df_spans_std=df_spans_std.rename(columns={'kepler_duration_sample':'kepler_duration_sample_std', 'kepler_energy_sample':'kepler_energy_sample_std',
                                            'energy_sample':'energy_sample_std','cpu_ratio':'cpu_ratio_std','energy_ratio':'energy_ratio_std'})

df_spans_summary = pd.DataFrame([df_spans_mean.cpu_ms,df_spans_mean.active_interval_ms,
                                 df_spans_mean.kepler_duration_sample_mean,df_spans_std.kepler_duration_sample_std,
                                df_spans_mean.kepler_energy_sample_mean, df_spans_std.kepler_energy_sample_std, 
                                df_spans_mean.energy_sample_mean, df_spans_std.energy_sample_std,
                                df_spans_mean.cpu_ratio_mean, df_spans_std.cpu_ratio_std,
                                df_spans_mean.energy_ratio_mean,df_spans_std.energy_ratio_std ]).transpose()
df_spans_summary
# df_mean = df_summary.groupby(['cpu_ms','active_interval_ms'], as_index=False).mean()
# df_mean=df_mean.rename(columns={'energy_joules':'mean_energy_joules'})
```
